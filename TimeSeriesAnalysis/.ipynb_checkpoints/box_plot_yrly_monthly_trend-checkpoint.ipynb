```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('data/milk.xlsx',header=None)
df.head()
```

```python
mseries = pd.concat([df.loc[0], df.loc[1], df.loc[2], df.loc[3], df.loc[4], df.loc[5], df.loc[6], df.loc[7], df.loc[8],
                    df.loc[9], df.loc[10], df.loc[11], df.loc[12], df.loc[13], df.loc[14], df.loc[15], df.loc[16]], ignore_index='True')
mts= mseries.drop([168,169])
```

```python
timeindex=pd.date_range(start='1962-01', periods=168, freq='ME')
timeindex
```

```python
mts.index = timeindex
mts.head()
```

```python
mts.plot()
plt.title("Milk Average per Cow from Jan.1962 to Dec.1975")
plt.xlabel("Year")
plt.ylabel("Milk Average")
plt.show()
```

```python
mts_reshape = np.array(mts).reshape(14,12)
mts_reshape
```

```python
mts_transpose = np.transpose(mts_reshape)
mts_transpose
```

```python
year = [1962, 1963, 1964, 1965, 1966, 1967, 1968, 1969, 1970, 1971,1972, 1973, 1974, 1975]
mts_df_year = pd.DataFrame(mts_transpose, columns = year)
mts_df_year.boxplot()
plt.xlabel('Year')
plt.ylabel('Milk Average')
plt.xticks(rotation=45)
```

```python
month=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul','Aug', 'Sep', 'Oct', 'Nov','Dec']
mts_df_month = pd.DataFrame(mts_reshape, columns = month)
mts_df_month.boxplot()
plt.xlabel('Year')
plt.ylabel('Milk Average')
plt.xticks(rotation=45)
```

```python
ftab = FreqTab.from_seq([1,2,2,3,5])
n = ftab.sum()
pmf = ftab/n
pmf
```

```python
from empiricaldist import Pmf

pmf = Pmf.from_seq([1,2,2,3,5])
pmf
```

```python
mean = np.sum(pmf.ps * pmf.qs) #expected value
mean
```

```python
pmf.mean()
```

```python
deviations = pmf.qs - mean
deviations
```

```python
var = np.sum(pmf.ps * deviations **2)
var
```

```python
ranges = pd.interval_range(start = 5, end = 50, freq = 5, closed = "left")
ranges.name = "class_size"
data = pd.DataFrame(index= ranges)
data["count"] = [8,8,14,4,6,12,8,3,2]
sizes = ranges.left + 2
n = data["count"].sum()
data["prob"] = data["count"]/n
s = data["prob"]*sizes
s.sum()
```

```python
counts = data["count"]
pmf_actual = Pmf(counts, sizes, name="actual")
pmf_actual.normalize()
pmf_actual.mean()
```

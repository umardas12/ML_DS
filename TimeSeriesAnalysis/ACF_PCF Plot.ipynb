```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import acf, pacf

df = pd.read_csv('data/gdpquarterlychina1992.1-2017.4.csv')
dates = pd.date_range(start='1992', periods=len(df), freq='QE')
df.index=dates
df.head()
```

```python
df.plot()
plt.title("Chinese Quaterly GDP year 1992 to 2017")
plt.ylabel("billions of RMB")
```

```python
acf_lag = acf(df, nlags = 60, fft = False)
plt.figure(figsize=(10,6))
plt.vlines(range(60+1),[0],acf_lag)
plt.axhline(y=0, lw=2)
plt.axhline(y=1.96/np.sqrt(len(df)), color='blue', linestyle = '--', linewidth = 1)
plt.axhline(y=-1.96/np.sqrt(len(df)), color='blue', linestyle = '--', linewidth = 1)
plt.xticks(ticks=range(0,60+1,3))
plt.xlabel('Lag')
plt.ylabel('ACF')
plt.tight_layout()
plt.show()
```

```python
pacf_lag = pacf(df, nlags = 30, method='ywmle')
plt.figure(figsize=(10,6))
plt.vlines(range(30+1),[0],pacf_lag)
plt.axhline(y=0, lw=2)
plt.axhline(y=1.96/np.sqrt(len(df)), color='blue', linestyle = '--', linewidth = 1)
plt.axhline(y=-1.96/np.sqrt(len(df)), color='blue', linestyle = '--', linewidth = 1)
plt.xticks(ticks=range(0,30+1,3))
plt.xlabel('Lag')
plt.ylabel('PACF')
plt.tight_layout()
plt.show()
```
